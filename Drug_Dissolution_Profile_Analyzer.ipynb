```python
import numpy as np

time = np.array([0, 5, 10, 15, 30, 45, 60])
formulation_A = np.array([0, 18, 35, 52, 72, 86, 94])
formulation_B = np.array([0, 12, 27, 43, 63, 78, 89])

DE_A = np.trapezoid(formulation_A, time) / 6000 * 100
DE_B = np.trapezoid(formulation_B, time) / 6000 * 100

def time_to_80(release):
    for i, value in enumerate(release):
        if value >= 80:
            return time[i]
    return "Not reached"

print("DISSOLUTION PROFILE ANALYSIS")
print("-" * 35)
print(f"Formulation A → DE: {DE_A:.2f}% | T80: {time_to_80(formulation_A)} min")
print(f"Formulation B → DE: {DE_B:.2f}% | T80: {time_to_80(formulation_B)} min")

print("\nInterpretation:")
print("Formulation A shows faster and greater drug release than Formulation B.")
print("The data are simulated for educational demonstration.")
```
